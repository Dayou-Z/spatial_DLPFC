```python
import scipy.io
import scanpy as sc
import pandas as pd
import os
```

```python
adata = sc.read_h5ad(os.path.expanduser(
    "../data/processed/02_preprocessed.h5ad"))
```

```python
print("counts" in adata.layers)
print("array_row" in adata.obs.columns)
print("array_col" in adata.obs.columns)
print(adata.shape)
```

```python
out_dir = os.path.expanduser("../data/processed")
os.makedirs(out_dir, exist_ok=True)

mat = adata.layers["counts"]

scipy.io.mmwrite(os.path.join(out_dir, "counts.mtx"), mat.T) 

pd.DataFrame(adata.var_names, columns=["gene"]).to_csv(
    os.path.join(out_dir, "genes.csv"), index=False
)

meta = adata.obs[["array_row", "array_col", "ground_truth"]].copy()
meta["barcode"] = adata.obs_names
meta.to_csv(os.path.join(out_dir, "metadata.csv"), index=False)
```
